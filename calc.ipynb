```python
from calc import BGMLVLE
import numpy as np
import pandas as pd
df = pd.read_excel('./test/test_newbinary.xlsx')
papers = df['paper'].unique().tolist()
VLE = BGMLVLE()
for paper in papers:
    # 获取paper的index
    index = df[df['paper'] == paper].index.tolist()
    # 获取paper的所有数据
    mol1 = df.loc[index[0], 'mol1']
    mol2 = df.loc[index[0], 'mol2']
    tag = df.loc[index[0], 'tag']
    T = df.loc[index, 'T'].values
    P = df.loc[index, 'P'].values
    x1 = df.loc[index, 'x1'].values
    y1 = df.loc[index, 'y1'].values
    if tag == "P":
        y, T_bubble = VLE.Isobaric(mol1, mol2, P.mean(), x1, normality_test = True)
        df.loc[index, 'y_pred'] = y[0]
        df.loc[index, 'y_high'] = y[1]
        df.loc[index, 'y_low'] = y[2]
        df.loc[index, 'T_pred'] = T_bubble[0]
        df.loc[index, 'T_high'] = T_bubble[1]
        df.loc[index, 'T_low'] = T_bubble[2]
        if mol1 == 'N-HEXANE' and mol2 == 'METHANOL':
            xL1, xL2, T1 = VLE.LLE_fug_average_P(mol1, mol2, P.mean())
            y1, T_bubble = VLE.Isobaric(mol1, mol2, P.mean(), np.array([xL1]), normality_test = False)
            # xL1 < df.loc[index, 'x1'] < xL2，找出这部分的index
            index_in_range = df[(df['x1'] > xL1) & (df['x1'] < xL2) & (df['paper'] == paper)].index.tolist()
            df.loc[index_in_range, 'y_pred'] = y1[0].item()
            df.loc[index_in_range, 'y_high'] = y1[1].item()
            df.loc[index_in_range, 'y_low'] = y1[2].item()
            df.loc[index_in_range, 'T_pred'] = T_bubble[0].item()
            df.loc[index_in_range, 'T_high'] = T_bubble[1].item()
            df.loc[index_in_range, 'T_low'] = T_bubble[2].item()
        df.loc[index, 'y_aard'] = 100 * np.abs((df.loc[index, 'y_pred'] - df.loc[index, 'y1']) / df.loc[index, 'y1'])
        df.loc[index, 'T_aard'] = 100* np.abs((df.loc[index, 'T_pred'] - df.loc[index, 'T']) / df.loc[index, 'T'])
    else:
        y, P_bubble = VLE.Isothermal(mol1, mol2, T.mean(), x1, normality_test = True)
        df.loc[index, 'y_pred'] = y[0]
        df.loc[index, 'y_high'] = y[1]
        df.loc[index, 'y_low'] = y[2]
        df.loc[index, 'P_pred'] = P_bubble[0]
        df.loc[index, 'P_high'] = P_bubble[1]
        df.loc[index, 'P_low'] = P_bubble[2]
        if mol1 == 'N-HEXANE' and mol2 == 'METHANOL':
            xL1, xL2 = VLE.LLE_fug_average(mol1, mol2, T.mean())
            y1, P_bubble = VLE.Isothermal(mol1, mol2, T.mean(), np.array([xL1]), normality_test = False, thermodynamic_consistency_test = False)
            # xL1 < df.loc[index, 'x1'] < xL2，找出这部分的index
            index_in_range = df[(df['x1'] > xL1) & (df['x1'] < xL2) & (df['paper'] == paper)].index.tolist()
            df.loc[index_in_range, 'y_pred'] = y1[0].item()
            df.loc[index_in_range, 'y_high'] = y1[1].item()
            df.loc[index_in_range, 'y_low'] = y1[2].item()
            df.loc[index_in_range, 'P_pred'] = P_bubble[0].item()
            df.loc[index_in_range, 'P_high'] = P_bubble[1].item()
            df.loc[index_in_range, 'P_low'] = P_bubble[2].item()
        df.loc[index, 'y_aard'] = 100 * np.abs((df.loc[index, 'y_pred'] - df.loc[index, 'y1']) / df.loc[index, 'y1'])
        df.loc[index, 'P_aard'] = 100* np.abs((df.loc[index, 'P_pred'] - df.loc[index, 'P']) / df.loc[index, 'P'])
# df.loc[index, 'y_aard']为NaN的行，赋值为0
df['y_aard'] = df['y_aard'].fillna(0)
df.to_excel('./test/test_newbinary_pred.xlsx', index=False)
```

### Temperature extrapolation

```python
from calc import BGMLVLE
import numpy as np
import pandas as pd

# 打开test.xlsx
df = pd.read_excel('./test/test_T_Extrapolation.xlsx')
papers = df['paper'].unique().tolist()
VLE = BGMLVLE()
for paper in papers:
    # 获取paper的index
    index = df[df['paper'] == paper].index.tolist()
    # 获取paper的所有数据
    mol1 = df.loc[index[0], 'mol1']
    mol2 = df.loc[index[0], 'mol2']
    tag = df.loc[index[0], 'tag']
    T = df.loc[index, 'T'].values
    P = df.loc[index, 'P'].values
    x1 = df.loc[index, 'x1'].values
    y1 = df.loc[index, 'y1'].values
    if tag == "P":
        y, T_bubble = VLE.Isobaric(mol1, mol2, P.mean(), x1, normality_test = True)
        df.loc[index, 'y_pred'] = y[0]
        df.loc[index, 'y_aard'] = 100* np.abs((y[0] - y1) / y1)
        df.loc[index, 'y_high'] = y[1]
        df.loc[index, 'y_low'] = y[2]
        df.loc[index, 'T_pred'] = T_bubble[0]
        df.loc[index, 'T_aard'] = 100* np.abs((T_bubble[0] - T) / T)
        df.loc[index, 'T_high'] = T_bubble[1]
        df.loc[index, 'T_low'] = T_bubble[2]
    else:
        y, P_bubble = VLE.Isothermal(mol1, mol2, T.mean(), x1, normality_test = True)
        df.loc[index, 'y_pred'] = y[0]
        df.loc[index, 'y_aard'] = 100* np.abs((y[0] - y1) / y1)
        df.loc[index, 'y_high'] = y[1]
        df.loc[index, 'y_low'] = y[2]
        df.loc[index, 'P_pred'] = P_bubble[0]
        df.loc[index, 'P_aard'] = 100* np.abs((P_bubble[0] - P) / P)
        df.loc[index, 'P_high'] = P_bubble[1]
        df.loc[index, 'P_low'] = P_bubble[2]
df['y_aard'] = df['y_aard'].fillna(0)
df.to_excel('./test/test_T_Extrapolation_pred.xlsx', index=False)
```
