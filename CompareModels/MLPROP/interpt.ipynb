```python
import numpy as np
import pandas as pd
import scipy
name = 'BENZENE_PERFLUOROBENZENE_P'
df = pd.read_csv(f'{name}.csv')
x = df['x_1'].values
Bubble = df['Bubble Line'].values
Dew = df['Dew Line'].values
# 构建x关于Bubble和Dew的样条插值函数
spline_bubble = scipy.interpolate.CubicSpline(x, Bubble)
spline_dew = scipy.interpolate.CubicSpline(x, Dew)
x_calc = [0,0.0638,0.1034,0.1156,0.1494,0.1526,0.195,0.2198,0.2336,0.261,0.2655,0.2771,0.2959,0.3174,0.3417,0.3828,0.5169,0.56,0.6246,0.6783,0.7274,0.7487,0.7594,0.77,0.8051,0.8466,0.8809,0.9089,0.9315,0.9494,0.96,0.9785,1]
bubble_calc = spline_bubble(x_calc)
# 用scipy反解Dew对应的x，以x_calc[i]作为初值
dew_calc = []
for x0 in x_calc:
    func = lambda x: spline_bubble(x0) - spline_dew(x)
    # 求根，不能用newton法，因为没有导数信息,使用单纯型
    root = scipy.optimize.root(func, x0=x0, method='lm')
    dew_calc.append(root.x[0])
for i in range(len(x_calc)):
    print(bubble_calc[i] * (1 if name[-1] != 'T' else 1e5))
```
